## Handling Missing Values

```python
df.dropna(subset=['Customer_ID'], inplace = True)
```

```python
df.isnull().sum()
```

```python
df['Age'].unique()
```

```python
import re 
def extract_age(age) :
    age_num=re.findall('[0-9]+',str(age))
    if len(age_num) >0  :
        return age_num[0]
    else :
        return age

df['Age'] = df['Age'].apply(lambda x : extract_age(x))
```

```python
df['Age'].unique()
```

```python
df_age = df[df['Age'] != 'nan years']['Age']
```

```python
age_median = int(df_age.dropna().astype('int64').median())
```

```python
age_median
```

```python
df = df.replace('nan years' , age_median )
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

age_median = int(df['Age'].median())

df['Age'] = df['Age'].fillna(age_median)

print(df['Age'].isnull().sum())
```

```python
df.isnull().sum()
```

```python
df['Purchase_Amount']=df['Purchase_Amount'].fillna(df['Purchase_Amount'].median() )
```

```python
df['Feedback_Score']=df['Feedback_Score'].fillna(df['Feedback_Score'].median()  )
```

```python
df.isnull().sum()
```

```python
for col in ['Gender', 'City', 'Country']:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df['Last_Purchase_Date'].ffill(inplace = True)
```

```python
df.isnull().sum()
```

```python
df['Gender'].unique()
```

```python
df['Gender']=df['Gender'].str.lower().str.strip()
```

```python
df['Gender'] = df['Gender'].replace({'m':'male','f':'female'})
```

```python
df['Gender'].unique()
```

```python
df['City'] = df['City'].str.lower().str.strip()
```

```python
df['City'].unique()
```

```python
df['Country']=df['Country'].str.lower()
```

```python
df['Country'].unique()
```

```python
df['Country'] = df['Country'].replace({'ind':'india'})
```

```python
df['Country'].unique()
```

```python
df['Country'] = df['Country'].fillna('india')
```

### handling outliers

```python
from scipy.stats import zscore
import numpy as np

z_score = np.abs(zscore(df[['Age','Purchase_Amount']]))
```

```python
df_clean = df[~(z_score>3).any(axis=1)]
```
